```python
import pandas as pd
import ast

# --- 1. Load Data ---
df_final = pd.read_csv('../../1_data_collection/extracting_company_data/general_info.csv')
print(f"Data loaded. Total rows: {len(df_final)}")

# --- 2. Prepare 'Domain' Column ---
if not df_final.empty and isinstance(df_final['domain'].iloc[0], str):
    df_final['domain'] = df_final['domain'].apply(lambda x: ast.literal_eval(x))
else:
    print("Domain column appears to contain lists or is empty. No string evaluation needed.")

# --- 3. Explode and Clean Domains ---
df_domain_table = df_final.explode('domain')
df_domain_table['domain'] = df_domain_table['domain'].str.replace(r'\.\.\.$', '', regex=True).str.strip()

# --- 4. Count Domain Frequency ---
domain_counts = df_domain_table['domain'].value_counts().reset_index()
domain_counts.columns = ['domain', 'count']

# --- 5. Generate domain_id in format "rank_count" ---
domain_counts['domain_id'] = domain_counts.reset_index().index + 1  # Rank
domain_counts['domain_id'] = domain_counts['domain_id'].astype(str) + '_' + domain_counts['count'].astype(str)

# --- 6. Final Domain Table ---
domain_df = domain_counts[['domain_id', 'domain']]

# --- 7. Create Company_Domain Table ---
company_domain_df = df_domain_table.merge(domain_df, on='domain', how='left')
company_domain_df = company_domain_df[['website', 'domain_id']]

# --- 8. Output ---
print("\n--- Domain Table ---")
print(domain_df.head())

print("\n--- Company_Domain Table ---")
print(company_domain_df.head())


```

```python

domain_df.to_csv('domain_table.csv', index=False)
company_domain_df.to_csv('company_domain_table.csv', index=False)
```
