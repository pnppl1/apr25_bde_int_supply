```python
import pandas as pd
import ast # For safe evaluation of string literals

# --- 1. Load Data ---
# Load the main company data from the CSV file.
df_final = pd.read_csv('general_info.csv')
print(f"Data loaded. Total rows: {len(df_final)}")

# --- 2. Prepare 'Domain' Column ---
# Ensure the 'domain' column contains actual Python lists.
# This handles cases where lists might be stored as strings (e.g., "['A', 'B']").
if not df_final.empty and isinstance(df_final['domain'].iloc[0], str):
    df_final['domain'] = df_final['domain'].apply(lambda x: ast.literal_eval(x))
else:
    print("Domain column appears to contain lists or is empty. No string evaluation needed.")

# --- 3. Explode and Clean Domains ---
# Explode the 'domain' column to create one row per individual domain.
# This duplicates company and website info for each domain.
df_domain_table = df_final.explode('domain')

# Clean up individual domain names by removing trailing '...' and whitespace.
df_domain_table['domain'] = df_domain_table['domain'].str.replace(r'\.\.\.$', '', regex=True).str.strip()

# --- 4. Select and Rename Columns ---
# Select desired columns: original company index, company name, website, and the single domain.
df_domain_table = df_domain_table[['company', 'website', 'domain']]
df_domain_table = df_domain_table.rename(columns={'index': 'original_company_index'})

# --- 5. Display Result ---

df_domain_table.head(20)
#print(f"Total rows in new 'Domain' table: {len(df_domain_table)}")
```
