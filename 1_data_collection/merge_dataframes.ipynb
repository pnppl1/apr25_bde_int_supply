```python
import pandas as pd
```

```python
general_info_file_path = "extracting_company_data/general_info.csv"
df_general_info = pd.read_csv(general_info_file_path)
display(df_general_info)
review_classes_file_path = "extracting_company_data/review_classes.csv"
df_review_classes = pd.read_csv(review_classes_file_path)
display(df_review_classes)
```

```python
df_companies = (
    df_general_info
    .merge(df_review_classes, how="left", on="website")
    )
display(df_companies)
display(df_companies.isna().sum())
```
